
#  Uber Trip Demand Analysis & Forecasting

## Professional EDA & Insights Report

---

## 1. Project Overview

This project analyzes Uber trip demand data to understand usage patterns, supply–demand relationships, and demand growth trends.
The analysis is based on real operational data from January–February 2015 and includes exploratory data analysis, feature engineering, and forecasting preparation.

---

## 2. Dataset Summary

* **Time Period:** January 2015 – February 2015
* **Granularity:** Daily trip-level aggregation
* **Key Variables:**

  * `trips` – Total Uber trips per day
  * `active_vehicles` – Number of active Uber vehicles
  * `date` – Date of operation

---

## 3. Supply vs Demand Analysis

### Average Uber Trips by Active Vehicle Level

| Active Vehicle Level | Average Trips |
| -------------------- | ------------- |
| Low                  | **1,653**     |
| Medium               | **7,140**     |
| High                 | **11,481**    |
| Very High            | **26,341**    |

### Insight

* When vehicle availability is **Low**, Uber completes only **~1.6k trips/day**
* With **Very High** vehicle availability, average trips increase to **~26.3k/day**
* This represents a **~16× increase** from Low to Very High levels

**Conclusion:**
Vehicle availability is a critical driver of realized demand. Increasing active vehicles directly enables higher trip completion.

---

## 4. Weekly Usage Pattern

### Average Uber Trips by Day of Week

| Day       | Average Trips |
| --------- | ------------- |
| Monday    | **9,662**     |
| Tuesday   | **10,063**    |
| Wednesday | **11,067**    |
| Thursday  | **12,326**    |
| Friday    | **13,170**    |
| Saturday  | **13,913**    |
| Sunday    | **10,915**    |

### Insight

* Demand steadily increases from **Monday → Saturday**
* **Saturday** records the highest demand (**13,913 trips/day**)
* Slight decline observed on **Sunday**

**Conclusion:**
Uber usage peaks on weekends, indicating strong leisure and non-work travel behavior.

---

## 5. Daily Demand Trend (Smoothed)

### 7-Day Smoothed Daily Uber Trips

* Daily trips fluctuate significantly due to weekday/weekend effects
* 7-day rolling average smooths noise and reveals a **clear upward trend**
* Demand grows steadily from early January to late February

**Conclusion:**
Uber demand shows consistent growth across the observed period, not driven by isolated spikes.

---

## 6. Monthly Demand Growth

### Monthly Uber Trips

| Month   | Total Trips   |
| ------- | ------------- |
| 2015-01 | **1,908,649** |
| 2015-02 | **2,221,581** |

### Growth Calculation

[
\text{Growth} = \frac{2,221,581 - 1,908,649}{1,908,649} \times 100
]

➡ **Monthly Growth = 16.4%**

### Insight

* Uber trips increased by **312,932 trips** month-over-month
* Indicates rapid platform adoption and increased usage
